```python
from selenium import webdriver
```

```python
import time
```

```python
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
```

---

## Open Facebook using Selenium

```python
browser = webdriver.Firefox()
browser.get("https://www.facebook.com/?locale=ar_AR")
time.sleep(3)

browser.find_element(By.NAME, "email").send_keys("***********")
time.sleep(3)

browser.find_element(By.NAME, "pass").send_keys("***********")
time.sleep(3)

browser.find_element(By.XPATH, "//span[text() = 'تسجيل الدخول']").click()
time.sleep(3)
```

---

## Navigate through the interface & Create a text post

```python
post_box = browser.find_element(By.XPATH, "//span[contains(text(), 'بم تفكر')]")
post_box.click()
time.sleep(3)


omar = browser.find_element(By.XPATH, "//div[@role='textbox' and @contenteditable='true']")
omar.click()
time.sleep(3)

text = "I’m taking the AI course with Eng.Ahmed Hafez"
for char in text:
    omar.send_keys(char)
    time.sleep(0.3)
```
