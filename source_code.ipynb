## Product urls: 

```python
browser = webdriver.Firefox()
browser.get("https://www.carrefouregypt.com/mafegy/ar/c/FEGY1660000")
time.sleep(5)

for i in range(11):
    browser.execute_script("window.scrollBy(0,600);")
    time.sleep(2)

all_urls = browser.find_elements(By.XPATH, "//span[contains(@class, 'line-clamp-2')]/..")
print(len(all_urls))

for element in all_urls:
    print("Url is: ", element.get_attribute("href"))

browser.quit()    
```

---

## Combine Data: 

```python
browser = webdriver.Firefox()
browser.get("https://www.carrefouregypt.com/mafegy/ar/c/FEGY1660000")
time.sleep(5) 


for i in range(11):
    browser.execute_script("window.scrollBy(0,600);")
    time.sleep(2)

all_names = browser.find_elements(By.CLASS_NAME, "line-clamp-2")
all_prices = browser.find_elements(By.CLASS_NAME, "items-baseline")
all_urls = browser.find_elements(By.XPATH, "//span[contains(@class, 'line-clamp-2')]/..")



product_list = []
for name, price, url in zip(all_names, all_prices, all_urls):
    product_dict = {
        "product_name" : name.text,
        "product_price" : price.text.replace("\n", " "),
        "product_url" : url.get_attribute("href"),
    }

    product_list.append(product_dict)

print(len(product_list))
print(product_list[:3])

# browser.quit()
```

---

## Save Data (CSV/JSON)

```python
import csv
```

```python
with open("carrefour_products.csv", "w", newline="", encoding="utf-8-sig") as file:
    writer = csv.DictWriter(file, fieldnames = ["product_name", "product_price", "product_url"])
    writer.writeheader()
    writer.writerows(product_list)

print("CSV Saved", len(product_list), "products")    
```

```python
import json
```

```python
with open("carrefour_products.json", "w", newline="", encoding="utf-8") as file:
    writer = json.dump(product_list, file, ensure_ascii=False, indent=4)

print("JSON Saved", len(product_list), "products")        
```

### Bonus: Discount Price

```python
browser = webdriver.Firefox()
browser.get("https://www.carrefouregypt.com/mafegy/ar/c/FEGY1660000")
time.sleep(5)

for i in range(11):
    browser.execute_script("window.scrollBy(0,600);")
    time.sleep(2)

all_prices = browser.find_elements(By.CLASS_NAME, "items-baseline")
print(len(all_prices))
print("----------")

all_names = browser.find_elements(By.CLASS_NAME, "line-clamp-2")

for name, price in zip(all_names, all_prices):
    old_price = price.find_elements(By.XPATH, ".//div[contains(@class, 'line-through')]")
    
    if len(old_price) > 0:
        new_price = price.find_elements(By.TAG_NAME, "span")
        print("Product Name is: ", name.text, " , ", "old price is:", old_price[0].text, " , ", "New Price is: ", new_price[1].text, new_price[0].text)

browser.quit()        
```
